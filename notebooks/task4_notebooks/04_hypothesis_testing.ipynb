```python
import pandas as pd
import scipy.stats as stats
```

```python
df = pd.read_csv('../data/processed/cleaned_data.csv')

# Create Sales column
df['Sales'] = df['Quantity'] * df['UnitPrice']
```

```python
# Total sales per customer
customer_sales = df.groupby('CustomerID')['Sales'].sum()

# Define threshold (median)
threshold = customer_sales.median()

# Create groups
high_value = customer_sales[customer_sales > threshold]
low_value = customer_sales[customer_sales <= threshold]
```

```python
t_stat, p_value = stats.ttest_ind(high_value, low_value, equal_var=False)

print("T-Statistic:", t_stat)
print("P-Value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject Null Hypothesis")
    print("High-value customers spend significantly more")
else:
    print("Fail to Reject Null Hypothesis")
```
